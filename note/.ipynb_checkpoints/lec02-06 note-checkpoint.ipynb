### Lecture 4 Pandas III

```python
#自定义排序 key
elections.sort_values(by="Candidate", key=lambda x : x.str.len(), ascending=False).head()
```

```python
#用map排序
def dr_ea_count(string):
    return string.count('dr') + string.count('ea')
elections["dr_ea_count"] = elections["Candidate"].map(dr_ea_count)
elections.sort_values(by="dr_ea_count", ascending=False)
```

```python
#dataframe.gruopby(column_name).agg(aggregation_function)
elections[["Year", "Popular vote"]].groupby("Year").agg(sum)
```

```python
elections.groupby("Year").sum(numeric_only=True)
```

```python
elections.drop("dr_ea_count", axis="columns", inplace=True)
elections.drop("Length", axis="columns", inplace=True)
elections
```

```python
elections.groupby("Year").filter(lambda sf : sf["%"].max() < 45)
```

```python
elections.sort_values(by="%", ascending=False).groupby("Party").first()
```

```python
elections.sort_values(by="%").drop_duplicates(["Party"], keep="last")
```

```python
grouped_by_party = elections.groupby("Party")
grouped_by_party.groups
```

```python
grouped_by_party.get_group("Socialist")
```

```python
elections_pivot = elections.pivot_table(
    index = "Year",
    columns = "Result",
    values = ["%"],
    aggfunc = np.max,
)
elections_pivot.head(5)
```

### Lecture 5 Data Wrangling and EDA (Part 1)

```python
tb_df = pd.read_csv("../lecture/lec05/data/cdc_tuberculosis.csv")
tb_df
```

```python
tb_df = pd.read_csv("../lecture/lec05/data/cdc_tuberculosis.csv", header=1) #表头第1行
tb_df
```

```python
rename_dict = {'2019' : 'TB cases 2019',
               '2020' : 'TB cases 2020',
               '2021' : 'TB cases 2021',
               '2019.1' : 'TB incidence 2019',
               '2020.1' : 'TB incidence 2020',
               '2021.1' : 'TB incidence 2021'}
tb_df = tb_df.rename(columns=rename_dict)
tb_df
```

```python
tb_df.drop(0)
```

```python
tb_df.dtypes
```

```python
tb_df = (
    pd.read_csv("../lecture/lec05/data/cdc_tuberculosis.csv",
                header=1,
                thousands=',') #1000里有逗号
        .rename(columns=rename_dict)
)

tb_df
```

```python
tb_df.dtypes
```

```python
tb_df.drop(0).sum()
```

```python
census_2010s_df = pd.read_csv("../lecture/lec05/data/nst-est2019-01.csv",
                              header=3,
                              thousands=',')
census_2010s_df = (
    census_2010s_df
    .rename(columns={"Unnamed: 0": "Geographic Area"})
    .drop(columns={"Census", "Estimates Base"})
    .convert_dtypes()
    .dropna()
)
census_2010s_df
```

```python
census_2010s_df['Geographic Area'] = census_2010s_df['Geographic Area'].str.strip('.')
census_2010s_df
```

```python
census_2020s_df = pd.read_csv("../lecture/lec05/data/NST-EST2022-POP.csv",
                              header=3,
                              thousands=',')
census_2020s_df = (
    census_2020s_df
    .drop(columns=["Unnamed: 1"])
    .rename(columns={"Unnamed: 0" : "Geographic Area"})
    .convert_dtypes()
    .dropna()
)
census_2020s_df['Geographic Area'] = census_2020s_df['Geographic Area'].str.strip('.')
census_2020s_df
```

```python
#show the three tables
display(tb_df.tail(2))
display(census_2010s_df.tail(2))
display(census_2020s_df.tail(2))
```

```python
#pd.merge join data on keys
tb_census_df = (
    tb_df
    .merge(right=census_2010s_df[["Geographic Area", "2019"]],
           left_on="U.S. jurisdiction", right_on="Geographic Area")
    .drop(columns="Geographic Area")
    .merge(right=census_2020s_df[["Geographic Area", "2020", "2021"]],
           left_on="U.S. jurisdiction", right_on="Geographic Area")
    .drop(columns="Geographic Area")
)
tb_census_df.tail()
```

```python
tb_census_df["recompute incidence 2019"] = (
    tb_census_df["TB cases 2019"]/tb_census_df["2019"] * 100000
)
tb_census_df
```

```python
for year in [2019, 2020, 2021]:
    tb_census_df[f"recompute incidence {year}"] = (
        tb_census_df[f"TB cases {year}"]/tb_census_df[f"{year}"]*100000
    )
tb_census_df
```

### Lecture 6 Data Wrangling and EDA (Part 2) & Text Wrangling and Regex

```python
states = pd.read_csv("../lecture/lec05/data/county_and_state.csv")
populations = pd.read_csv("../lecture/lec05/data/county_and_population.csv")
display(states.head())
display(populations.head())
```

```python
#不对表格进行规范化，merge会返回无结果
def canonicalize_county(county_series):
    return (county_series
            .str.lower()
            .str.replace(' ', '')
            .str.replace('&', 'and')
            .str.replace('.', '')
            .str.replace('county', '')
            .str.replace('parish', '')
           )

display(canonicalize_county(states["County"]))
display(canonicalize_county(populations["County"]))
```

```python
#正则表达式
# * 零次或多次
# + 一次或多次
# ？ 零次或一次
# {x, y} 最少x次，最多y次
# [A-Za-z0-9] 范围
# . 表示任意字符
# *? 表示*不再贪婪
# \ 读取下一个字符
# [^] 表示非 not
# ^ 表示开头
# $ 表示结尾
import re
# re.findall(pattern, text) return a list of all matches
```

```python
# 用()形成捕获组 str.extract str.extractall
pattern_cg = r"([0-9]{3})-([0-9]{2})-([0-9]{4})"
re.findall("123-45-6789", pattern_cg)
```
