```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
wb = pd.read_csv("../lecture/lec07/data/world_bank.csv", index_col=0)
wb.head()
```

```python
wb.shape
```

```python
wb["Continent"].value_counts()
```

```python
continents = wb["Continent"].value_counts()
continents
```

```python
plt.bar(continents.index, continents.values)
plt.xlabel("Continents")
plt.ylabel("Frequency")
plt.title("Frequency of countries in each continent");
```

```python
wb["Continent"].value_counts().plot(kind = "bar")
plt.xlabel("Continents")
plt.ylabel("Frequency")
plt.title("Frequency of countries in each continent");
```

```python
sns.countplot(data=wb, y="Continent");
```

```python
sns.boxplot(data = wb, y = 'Gross domestic product: % growth : 2016');
#箱型图 四分位距 IQR = Q3 - Q1
#上界是 Q3+1.5*IQR 下界是 Q1-1.5*IQR
```

```python
sns.violinplot(data=wb, y='Gross domestic product: % growth : 2016')
```

```python
sns.boxplot(data=wb, x="Continent", y='Gross domestic product: % growth : 2016')
```

```python
sns.histplot(data=wb, x='Gross domestic product: % growth : 2016', stat='density')
```
